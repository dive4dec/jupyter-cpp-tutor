## Simple Variables

## For Loop

```python
%%cpptutor
int main() {
    int sum = 0;
    for (int i = 1; i <= 5; i++) {
        sum += i;
    }
    return 0;
}
```

## While Loop

```python
%%cpptutor
int main() {
    int n = 10;
    int count = 0;
    while (n > 1) {
        n /= 2;
        count++;
    }
    return 0;
}
```

## Function Call

```python
%%cpptutor
int add(int a, int b) {
    return a + b;
}
int main() {
    int x = add(3, 4);
    int y = add(x, 10);
    return 0;
}
```

## Recursion: Euclidean GCD

```python
%%cpptutor
int gcd(int a, int b) {
    if (b == 0) return a;
    return gcd(b, a % b);
}
int main() {
    int result = gcd(48, 18);
    return 0;
}
```

## Recursion: Factorial

```python
%%cpptutor
int factorial(int n) {
    if (n <= 1) return 1;
    return n * factorial(n - 1);
}
int main() {
    int result = factorial(5);
    return 0;
}
```

## Pointer and Reference

```python
%%cpptutor
int main() {
    int x = 42;
    int *p = &x;
    int &r = x;
    *p = 100;
    r = 200;
    return 0;
}
```

## Heap Allocation

```python
%%cpptutor
int main() {
    int *p = new int(42);
    *p = 100;
    delete p;
    return 0;
}
```

## Array

```python
%%cpptutor
int main() {
    int arr[5] = {10, 20, 30, 40, 50};
    int sum = 0;
    for (int i = 0; i < 5; i++) {
        sum += arr[i];
    }
    return 0;
}
```

## Struct

```python
%%cpptutor
struct Point {
    int x;
    int y;
};
int main() {
    Point p;
    p.x = 3;
    p.y = 4;
    return 0;
}
```

## Polymorphism

```python
%%cpptutor
struct Shape {
    virtual int area() { return 0; }
};
struct Square : Shape {
    int side;
    Square(int s) : side(s) {}
    int area() override { return side * side; }
};
int main() {
    Square sq(5);
    Shape *s = &sq;
    int a = s->area();
    return 0;
}
```
