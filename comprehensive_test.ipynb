# jupyter-cpp-tutor: Comprehensive Test

This notebook tests all visualization features against Python Tutor's behavior.

**Key behaviors matched with pythontutor.com:**
- Function-entry steps show `int main() {` / `int add() {` with uninitialized vars (`?`)
- Variables show `?` before their assignment line executes
- Closing brace `}` of `main()` is NOT shown; `}` of other functions IS shown
- Heap objects are removed after `delete` executes
- Dual highlighting: green = just executed, yellow = next to execute


```python
%load_ext jupyter_cpp_tutor
```

## 1. Simple Variables

```python
%%cpptutor
int main() {
    int x = 1;
    int y = x + 2;
    return 0;
}
```

## 2. For Loop

```python
%%cpptutor
int main() {
    int sum = 0;
    for (int i = 1; i <= 3; i++) {
        sum += i;
    }
    return 0;
}
```

## 3. Function Call

```python
%%cpptutor
int add(int a, int b) {
    return a + b;
}
int main() {
    int x = add(3, 4);
    return 0;
}
```

## 4. Pointer

```python
%%cpptutor
int main() {
    int x = 42;
    int *p = &x;
    return 0;
}
```

## 5. Array

```python
%%cpptutor
int main() {
    int arr[3] = {10, 20, 30};
    return 0;
}
```

## 6. Struct

```python
%%cpptutor
struct Point {
    int x;
    int y;
};
int main() {
    Point p;
    p.x = 3;
    p.y = 4;
    return 0;
}
```

## 7. Heap Allocation

```python
%%cpptutor
int main() {
    int *p = new int(42);
    *p = 100;
    delete p;
    return 0;
}
```

## 9. Configurable Std

```python
%%cpptutor
int main() {
    auto x = 42;
    return 0;
}
```

## 10. Recursion: Euclidean GCD

```python
%%cpptutor
int gcd(int a, int b) {
    if (b == 0) return a;
    return gcd(b, a % b);
}
int main() {
    int result = gcd(48, 18);
    return 0;
}
```
