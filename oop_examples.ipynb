## 1. Constructor and Destructor

```python
%%cpptutor
#include <iostream>
struct Widget {
    int id;
    Widget(int i) : id(i) {
        std::cout << "Widget " << id << " created" << std::endl;
    }
    ~Widget() {
        std::cout << "Widget " << id << " destroyed" << std::endl;
    }
};
int main() {
    Widget w1(1);
    Widget w2(2);
    return 0;
}
```

## 2. Member Functions

```python
%%cpptutor
#include <iostream>
struct Counter {
    int count;
    Counter() : count(0) {}
    void increment() { count++; }
    int getValue() { return count; }
};
int main() {
    Counter c;
    c.increment();
    c.increment();
    c.increment();
    int val = c.getValue();
    return 0;
}
```

## 3. Virtual Functions (Polymorphism)

```python
%%cpptutor
#include <iostream>
struct Shape {
    virtual int area() { return 0; }
};
struct Square : Shape {
    int side;
    Square(int s) : side(s) {}
    int area() override { return side * side; }
};
struct Rectangle : Shape {
    int w, h;
    Rectangle(int w, int h) : w(w), h(h) {}
    int area() override { return w * h; }
};
int main() {
    Shape *s1 = new Square(5);
    Shape *s2 = new Rectangle(3, 4);
    int a1 = s1->area();
    int a2 = s2->area();
    delete s1;
    delete s2;
    return 0;
}
```

## 4. Constructor Chaining (Inheritance)

```python
%%cpptutor
#include <iostream>
struct Base {
    int x;
    Base(int x) : x(x) {
        std::cout << "Base(" << x << ")" << std::endl;
    }
    virtual ~Base() {
        std::cout << "~Base" << std::endl;
    }
};
struct Derived : Base {
    int y;
    Derived(int x, int y) : Base(x), y(y) {
        std::cout << "Derived(" << x << "," << y << ")" << std::endl;
    }
    ~Derived() override {
        std::cout << "~Derived" << std::endl;
    }
};
int main() {
    Derived d(10, 20);
    return 0;
}
```

## 5. Diamond Problem (Virtual Inheritance)

```python
%%cpptutor
#include <iostream>
struct A {
    int a_val;
    A() : a_val(1) {
        std::cout << "A()" << std::endl;
    }
};
struct B : virtual A {
    int b_val;
    B() : b_val(2) {
        std::cout << "B()" << std::endl;
    }
};
struct C : virtual A {
    int c_val;
    C() : c_val(3) {
        std::cout << "C()" << std::endl;
    }
};
struct D : B, C {
    int d_val;
    D() : d_val(4) {
        std::cout << "D()" << std::endl;
    }
};
int main() {
    D d;
    int total = d.a_val + d.b_val + d.c_val + d.d_val;
    return 0;
}
```

## 6. Recursion: Euclidean GCD

```python
%%cpptutor
int gcd(int a, int b) {
    if (b == 0) return a;
    return gcd(b, a % b);
}
int main() {
    int result = gcd(48, 18);
    return 0;
}
```

## 7. Recursion: Factorial

```python
%%cpptutor
int factorial(int n) {
    if (n <= 1) return 1;
    return n * factorial(n - 1);
}
int main() {
    int result = factorial(5);
    return 0;
}
```
